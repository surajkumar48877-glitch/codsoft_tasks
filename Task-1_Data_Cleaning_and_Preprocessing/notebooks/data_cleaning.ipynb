## 7. Create the Dirty Practice Dataset

The practice copy contains:
- 4 missing values
- 3 exact duplicate records
- inconsistent capitalization
- extra spaces in categorical values


```python
dirty_df = df.copy()

# Missing values
dirty_df.loc[10, "Age"] = np.nan
dirty_df.loc[25, "Age"] = np.nan
dirty_df.loc[50, "Review Rating"] = np.nan
dirty_df.loc[75, "Review Rating"] = np.nan

# Exact duplicate records
duplicate_rows = dirty_df.iloc[[100, 200, 300]].copy()
dirty_df = pd.concat([dirty_df, duplicate_rows], ignore_index=True)

# Inconsistent categorical entries
dirty_df.loc[400, "Gender"] = " male "
dirty_df.loc[500, "Gender"] = "FEMALE"
dirty_df.loc[600, "Category"] = " clothing "
dirty_df.loc[700, "Category"] = "FOOTWEAR"

print("Dirty dataset shape:", dirty_df.shape)
```

## 8. Identify Data Quality Issues

```python
print("Missing values:")
display(dirty_df.isnull().sum()[dirty_df.isnull().sum() > 0])

print("\nDuplicate records:", dirty_df.duplicated().sum())

print("\nGender examples:", dirty_df.loc[[400, 500], "Gender"].tolist())
print("Category examples:", dirty_df.loc[[600, 700], "Category"].tolist())
```

## 9. Handle Missing Values

```python
age_median = dirty_df["Age"].median()
rating_median = dirty_df["Review Rating"].median()

dirty_df["Age"] = dirty_df["Age"].fillna(age_median)
dirty_df["Review Rating"] = dirty_df["Review Rating"].fillna(rating_median)

print("Age median used:", age_median)
print("Review Rating median used:", rating_median)
print("Total missing values after imputation:", dirty_df.isnull().sum().sum())
```

## 10. Remove Duplicate Records

```python
duplicates_before = dirty_df.duplicated().sum()
dirty_df = dirty_df.drop_duplicates()
duplicates_after = dirty_df.duplicated().sum()

print("Duplicates before:", duplicates_before)
print("Duplicates after:", duplicates_after)
print("Shape after duplicate removal:", dirty_df.shape)
```

## 11. Correct Inconsistent Categorical Values

```python
dirty_df["Gender"] = dirty_df["Gender"].str.strip().str.title()
dirty_df["Category"] = dirty_df["Category"].str.strip().str.title()

print("Gender values:", dirty_df["Gender"].unique())
print("Category values:", dirty_df["Category"].unique())
```

## 12. Correct and Validate Data Types

```python
dirty_df["Age"] = dirty_df["Age"].astype(int)
dirty_df.info()
```

## 13. Before vs After Summary

```python
before = {"Rows": 3903, "Columns": 19, "Missing Values": 4, "Duplicate Records": 3}
after = {
    "Rows": len(dirty_df),
    "Columns": dirty_df.shape[1],
    "Missing Values": int(dirty_df.isnull().sum().sum()),
    "Duplicate Records": int(dirty_df.duplicated().sum())
}
comparison = pd.DataFrame([before, after], index=["Before Cleaning", "After Cleaning"])
display(comparison)
```

## 14. Visualize Cleaning Results

```python
metrics = ["Missing Values", "Duplicate Records"]
before_values = [before[m] for m in metrics]
after_values = [after[m] for m in metrics]

x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(8, 5))
plt.bar(x - width/2, before_values, width, label="Before Cleaning")
plt.bar(x + width/2, after_values, width, label="After Cleaning")
plt.xticks(x, metrics)
plt.ylabel("Count")
plt.title("Data Quality Before vs After Cleaning")
plt.legend()
plt.tight_layout()
plt.show()
```

## 15. Final Validation

```python
print("========== FINAL VALIDATION ==========")
print("Dataset shape:", dirty_df.shape)
print("Total missing values:", dirty_df.isnull().sum().sum())
print("Total duplicate records:", dirty_df.duplicated().sum())
print("Age dtype:", dirty_df["Age"].dtype)
print("Review Rating dtype:", dirty_df["Review Rating"].dtype)
```
